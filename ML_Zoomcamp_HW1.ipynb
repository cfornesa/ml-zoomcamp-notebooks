```python
# Read the car_fuel_efficiency.csv file
df = pd.read_csv('car_fuel_efficiency.csv')
```

```python
records = len(df)
print(f"Q2. There are {records} records in the dataset.")
```

```python
df.columns
```

```python
# Get unique "fuel_type" column values
unique_fuel_types = df['fuel_type'].unique()
print(f"Q3. There are {len(unique_fuel_types)} unique fuel types in the dataset.")
```

```python
missing_values_columns = []

for col in df.columns:
  # See if the column has missing values
  if df[col].isnull().any():
    missing_values_columns.append(col)

print(f"Q4. There are {len(missing_values_columns)} columns with missing values.")
```

```python
# Get the maximum fuel efficiency of cars from Asia
max_fuel_efficiency = df[df['origin'] == 'Asia']['fuel_efficiency_mpg'].max()
print(f"Q5. The maximum fuel efficiency of cars from Asia is {max_fuel_efficiency:.1f}.")
```

```python
# Display the top fuel efficiency values for cars from Asia
asia_fuel_efficiency = df[df['origin'] == 'Asia']['fuel_efficiency_mpg'].sort_values(ascending=False)
display(asia_fuel_efficiency.head(10))
print(f"The maximum fuel efficiency from Asia in the filtered data is: {asia_fuel_efficiency.max():.1f}")
```

```python
# Get the number of missing value in the "horsepower" column
missing_values_horsepower = df['horsepower'].isnull().sum()
print(f"There are {missing_values_horsepower} missing values in the horsepower column.")
```

```python
horsepower_median = df['horsepower'].median()
print(f"The horse power median is {horsepower_median}.")
horsepower_mode = df['horsepower'].mode()
```

```python
# Use fillna to fill in the missing values with the mode
df['horsepower'] = df['horsepower'].fillna(horsepower_mode.iloc[0])
missing_values_horsepower = df['horsepower'].isnull().sum()
print(f"There are {missing_values_horsepower} missing values in the horsepower column.")
horsepower_median_after_fillna = df['horsepower'].median()
print(f"Q6. The horse power median after fillna is {horsepower_median_after_fillna}.")
```

```python
# Select all cars from Asia
asia_cars = df[df['origin'] == 'Asia']

# Select only the vehicle_weight and model_year columns
asia_cars = asia_cars[['vehicle_weight', 'model_year']]

# Select only the first 7 values
asia_cars = asia_cars.head(7)

# Get the underlying NumPy array
x = np.asarray(asia_cars.values)

# New y array
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])

# Compute w using the least squares formula: w = (X.T @ X)^-1 @ X.T @ y
XTX_inv_XT = np.linalg.inv(x.T @ x) @ x.T
w = XTX_inv_XT @ y

# Get the sum of w
w_sum = w.sum()
print(f"Q7. The sum of the elements of w is {w_sum}.")
```
